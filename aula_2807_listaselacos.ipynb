```python
#verificar valor de uma comanda

valor = float(input("Informe o valor a pagar:"))

if valor < 10:
  print("o valor a pagar é menor que 10")
else:
  print("o valor a pagar é 10 ou maior")
```

```python
# agora queremos que o usuario informe todos os valores de uma comanda

#Uso do .lower para a resposta ficar em minusculo
#caso queira colocar a resposta em maiusculo é .upper

saldo = 0

while True:
  valor = float(input("informe o valor do item: "))
  saldo = saldo + valor

  pergunta = input("tem um novo valor a informar?")
  if pergunta.lower() == "sim" or pergunta.lower == "s":
    print("informe o novo valor: ")
  else:
    print(f"o seu saldo é de {saldo}")
    break
```

```python
#cadastrar diversos produtos usando lista

list_produtos = []

while True:
  produto = input("informe o nome do produto: \n")
  list_produtos.append(produto)

  pergunta = input("tem um novo produto a informar? \n")
  if pergunta.lower() == "sim" or pergunta.lower() == "s":
    print("informe o novo produto: \n")
  else:
    for i in list_produtos:
      print(f"\n{i}")
    break
```

```python
#OBJETIVO - CADASTRAR PRODUTOS E VALORES PARA MOSTRAR A COMANDA FINAL

# --- CRIANDO LISTA VAZIA DE PRODUTOS E VALORES
list_produtos = []
list_valores = []

# --- CADASTRO DAS INFORMAÇÕES
while True:
  produto = input("Qual o produto que deseja inserir? ")
  valor = float(input(f"Informe o valor da(o) {produto} R$ "))

# --- INSERINDO OS VALORES NA LISTA
  list_produtos.append(produto)
  list_valores.append(valor)

# --- CONDIÇÃO PARA VER SE HÁ MAIS PRODUTOS PARA INSERIR
  pergunta = input("deseja cadastrar um novo produto? (sim ou não): ")
  print(" ")
  if pergunta.lower() == "sim" or pergunta.lower() == "s":
    print(" ")

# --- CASO NÃO, SOMAR AS CONTAS
  else:
  # i - produto e j - valor
  #para exibir mais de um item coloca .zip para compactar os itens
      for i,j in zip(list_produtos, list_valores):
        #produto -- R$ valor
        print(f"o produto {i} custa R$ {j}")
      print(f"\n\ntotal da compra R$ {sum(list_valores)}")
      taxa = sum(list_valores)*0.1
      print(f"valor com taxa de serviço de 10% R$ {round(sum(list_valores,taxa),2)}")
      break
```

```python
#JOGO FIZZBUZZ SE O NÚMERO FOR MULTIPLO DE 3 APARECE FIZZ SE FOR DE 5 APARECE BUZZ SE FOR DOS DOIS APARECE FIZZBUZZ
while True:

  n = int(input("Me dê um número limite: "))

  for i in range(1, n + 1):
      if i % 3 == 0 and i % 5 == 0:
          print("fizz buzz")
      elif i % 3 == 0:
          print("fizz")
      elif i % 5 == 0:
          print("buzz")
      else:
          print(i)
  pergunta = input("Quer continuar jogando? ")
  if pergunta.lower() == "sim" or pergunta.lower() == "s":
    continue
  else:
     print("bom jogo")
  break
```
