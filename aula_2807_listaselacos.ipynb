```python
# verificar tamanho da conta a pagar

valor = float(input("Informe o valor a pagar:"))

if valor < 10:
  print("o valor a pagar é menor que 10")
else:
  print("o valor a pagar é 10 ou maior")
```

```python
# agora queremos que o usuario indorme todos os valores de uma comanda

#o .lower faz um texto ficar minusculo, porque o usuario pode escrever maiusculo e o pc n entender

saldo = 0

while True:
  valor = float(input("informe o valor do item: "))
  saldo = saldo + valor

  pergunta = input("tem um novo valor a informar?")
  if pergunta.lower() == "sim" or pergunta.lower == "s":
    print("informe o novo valor: ")
  else:
    print(f"o seu saldo é de {saldo}")
    break
```

```python
#cadastrar diversos produtos

list_produtos = []

while True:
  produto = input("informe o nome do produto: \n")
  list_produtos.append(produto)

  pergunta = input("tem um novo produto a informar? \n")
  if pergunta.lower() == "sim" or pergunta.lower() == "s":
    print("informe o novo produto: \n")
  else:
    for i in list_produtos:
      print(i)
    break
```
